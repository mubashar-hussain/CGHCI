## Problem 1: Thresholding

Turn each grayscale value into black (`0`) or white (`255`).





<br>

##   Step 1 : Inputs and Output




**Input :** A grayscale image and a threshold value.

 **Output:** A black-and-white image containing only `0` and `255`.



<br>

##  Step 2 : Rule


If:

`pixel >= threshold`

then:

`output = 255`

Otherwise:

`output = 0`

The threshold value itself is treated as white.

<br>

##  Step 3 : Small Example



For threshold = `128`:

Input:

40   120   180
200  190   220

Applying the rule:

40 < 128 → 0
120 < 128 → 0
180 >= 128 → 255

200 >= 128 → 255
190 >= 128 → 255
220 >= 128 → 255

Output:

0   0   255
255   255   255


<br>

##  Step 4 : Python

```python
import numpy as np


def threshold_image(image: np.ndarray, threshold: int) -> np.ndarray:
    """Convert a grayscale image to black and white."""
    return np.where(image >= threshold, 255, 0).astype(np.uint8)


problem_1_input = np.array(
    [[40, 120, 180],
     [200, 190, 220]],
    dtype=np.uint8
)

problem_1_expected = np.array(
    [[0, 0, 255],
     [255, 255, 255]],
    dtype=np.uint8
)

np.testing.assert_array_equal(
    threshold_image(problem_1_input, 128),
    problem_1_expected
)

print("Problem 1 passed.")
```

#  Problem 2: Image Memory

Calculate how many bytes an uncompressed image uses.

<br>

### Step 1 : Inputs and Output

**Input:** Image width, image height, and bits per pixel (`bpp`).

**Output:** Image memory in bytes.

<br>

### Step 2 : Rule

The memory is calculated using: 

`bytes = width × height × bpp / 8`

We divide by 8 because there are 8 bits in one byte.

<br>

### Step 3 : Small Example

For an image of `1920 × 1080` with `24 bpp`:

`1920 × 1080 × 24 / 8`

`= 6,220,800 bytes`

Therefore, the image requires **6,220,800 bytes**.

<br>

##  Step 4 : Python


```python
def display_memory_bytes(width: int, height: int, bpp: int) -> int:
    """Return image memory in bytes."""
    return width * height * bpp // 8


expected_bytes = 1920 * 1080 * 24 // 8

assert display_memory_bytes(1920, 1080, 24) == expected_bytes

print("Problem 2 passed.")
print("Memory:", display_memory_bytes(1920, 1080, 24), "bytes")
```

#  Problem 3: Average 9 Pixels

Find the average value of a small `3 × 3` image block.

<br>

### Step 1 : Inputs and Output

**Input:** Nine pixel values arranged in a `3 × 3` array.

**Output:** One rounded average value.

<br>

### Step 2 : Rule

The average is:   

`average = sum of all 9 pixel values / 9`

<br>

### Step 3 : Small Example

Input:

10   20   10  
30   50   30  
10   20   10  

Sum = 190

Average:

`190 / 9 = 21.11`

Rounded average = **21**

<br>

##  Step 4 : Python


```python
def mean_filter_3x3(neighborhood: np.ndarray) -> int:
    """Return the rounded average of 9 pixels."""
    return int(round(np.sum(neighborhood, dtype=np.uint32) / 9))


problem_3_input = np.array(
    [[20, 30, 40],
     [40, 50, 40],
     [20, 30, 20]],
    dtype=np.uint8
)

assert mean_filter_3x3(problem_3_input) == 32

print("Problem 3 passed.")
print("Average:", mean_filter_3x3(problem_3_input))
```

#  Problem 4: Change Image Contrast

      Change image values from one range to another range.



<br>

### Step 1 : Inputs and Output

**Input:** An array of values, an input range, and an output range.

**Output:** An array containing values mapped to the new range.



<br>

### Step 2 : Rule

The formula is:

`new = (value - old_min) / (old_max - old_min) × (new_max - new_min) + new_min`

The result is clipped so that it remains inside the output range.



<br>

### Step 3 : Small Example

   Change `[50, 100, 150]` from the range `[50, 150]` to `[0, 255]`.

        50 → 0  
        100 → 127.5  
        150 → 255  

    Expected output:

`      [0, 127.5, 255]`

<br>

##  Step 4 : Python


```python
def contrast_stretch(
        image: np.ndarray,
        in_min: float,
        in_max: float,
        out_min: float = 0.0,
        out_max: float = 255.0,
) -> np.ndarray:
    """Change image values from one range to another."""
    result = (
            (image - in_min)
            / (in_max - in_min)
            * (out_max - out_min)
            + out_min
    )

    return np.clip(result, out_min, out_max)


problem_4_input = np.array(
    [30, 90, 180],
    dtype=np.float32
)

expected = np.array(
    [0.0, 102.0, 255.0],
    dtype=np.float32
)

np.testing.assert_allclose(
    contrast_stretch(problem_4_input, 30, 180),
    expected
)

print("Problem 4 passed.")
print("Output:", contrast_stretch(problem_4_input, 30, 180))
```

# 🔹 Problem 5 : Find an Edge with Sobel

      Use a Sobel filter to find the strength of an edge in the middle of a `3 × 3` grayscale block.

<br>

### Step 1 : Inputs and Output

  **Input:** A `3 × 3` grayscale image block.

   **Output:** Three values:

   - `gx` – horizontal edge response
   - `gy` – vertical edge response
   - `edge strength` – overall edge strength

<br>

### Step 2 : Rule

The Sobel kernels are:

 Gx:

-1   0   1  
-2   0   2  
-1   0   1  

 Gy:

-1  -2  -1  
 0   0   0  
 1   2   1  

 Multiply the image block by each kernel and add the results.

Finally:

`strength = sqrt(gx² + gy²)`


<br>

###  Step 3 : Small Example

Input:

20   20   200  
20   20   200  
20   20   200  

  For this block:

` gx = 720`

`gy = 0`

 Therefore:

`strength = sqrt(720² + 0²)`

`strength = 720`

So the edge strength is **720**.

<br>

##  Step 4 : Python


```python
def sobel_response(block: np.ndarray) -> tuple[float, float, float]:
    """Return gx, gy, and edge strength."""
    
    gx_kernel = np.array(
        [[-1, 0, 1],
         [-2, 0, 2],
         [-1, 0, 1]],
        dtype=np.float32
    )

    gy_kernel = np.array(
        [[-1, -2, -1],
         [0, 0, 0],
         [1, 2, 1]],
        dtype=np.float32
    )

    gx = float(np.sum(block * gx_kernel))
    gy = float(np.sum(block * gy_kernel))

    strength = float(np.sqrt(gx * gx + gy * gy))

    return gx, gy, strength


problem_5_input = np.array(
    [[20, 20, 200],
     [20, 20, 200],
     [20, 20, 200]],
    dtype=np.float32
)

gx, gy, strength = sobel_response(problem_5_input)

assert gx == 720.0
assert gy == 0.0
assert strength == 720.0

print("Problem 5 passed.")
print("gx =", gx)
print("gy =", gy)
print("Edge strength =", strength)
```
